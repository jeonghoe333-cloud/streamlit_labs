```python
import pandas as pd

data = {
    "height": [165, 172, None, 180, 158, 175],
    "weight": [55, 70, 65, 80, 50, 68],
    "age": [25, 30, 28, 35, 22, 29],
    "gender": ["F", "M", "M", "M", "F", "M"],
}

df = pd.DataFrame(data)
```

```python
# 결측치 채우기 : .fillna()
# 숫자형들을 평균값으로 채우기 - 키(height)
df['height'].fillna(df['height'].mean())
```

```python
df3 = df.copy() # 원본데이터는 왠만하면 보관하고, 카피본을 만드는게 나음
df3['height'] = df['height'].fillna(df['height'].mean())
df3
```

```python
# 결측치 채우기 : .fillna()
# 이번에는 범주형들을 최빈값으로 채우기 - 성별(gender)
df['gender'].mode() # 아~ 최빈값은 M 남자구나~!
type(df['gender'].mode()) # pandas.Series구나~ (파이썬에서 리스트가 판다스에선 1차형 시리즈라고 함ㅇㅇ) 엥...?왜...?
df['gender'].mode()[0] # 아 지금 다시보니까 0  M이였네?
type(df['gender'].mode()[0]) # M이 str이였구나...!
```

## 인코딩(Encoding)
: 문자 데이터를 숫자로 바꾸는 과정

- 레이블 인코딩 (Label Encoding)
  : 값에 숫자를 직접 매긴다 ex. 월화수목금토일
- 원-핫 인코딩 (One-hot Encoding)
  : 순서가 없고 참/거짓만 있는 경우 ex. 남녀

```python
# 레이블 인코딩
df['gender_num'] = df['gender'].map({'F':0, 'M':1})
df
```

```python
# 원-핫 인코딩
df_encoded = pd.get_dummies(df, columns=['gender']) # .get_dummies() 정말 많이 씁니다
df_encoded
```

## 스케일링
: 숫자 값의 범위를 조정
ex. 나이 -> 20~70, 연봉 -> 30,000,000 이 둘의 숫자차이가 너무 크다

- MinMaxScaler
: 사이킷런 안에 있는 함수 / 값을 0~1사이 범위로 바꿔준다
- StandardScaler
: 사이킷런 안에 있는 함수 / 평균이 0, 표준편차가 1이 되도록 조정 (=정규분포..?)

```python
# uv pip list를 보면 skitlearn이 없네요? python.org 홈페이지 - PyPI(외부라이브러리 확인)
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler() # 붕어빵.. 이 객체에 모든 MinMaxScaler의 기능을 넣어요?
df[['height_scaled']] = scaler.fit_transform(df[['height']])
df
# 키가 젤 작은 애가 0.000 젤 큰 애가 1.000
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df[['height_standard']] = scaler.fit_transform(df[['height']])
df
```
